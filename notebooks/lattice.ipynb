```python
import mlx.core as mx
import matplotlib.pyplot as plt
import hyperzeta as hz

%matplotlib inline
%config InlineBackend.figure_format = "retina"
%load_ext autoreload
%autoreload 2
```

```python
basis = mx.array([[1.0, 0.2], [0.4, 1.2]]).T
lattice1 = hz.Lattice(n_max=3, nd=2)
lattice2 = hz.Lattice(n_max=3, nd=2, basis=basis)
```

```python
print(lattice2.covolume)
print(lattice2.gram)
```

```python
plt.scatter(lattice1.n[:, 0], lattice1.n[:, 1], marker=".")
plt.scatter(lattice2.n[:, 0], lattice2.n[:, 1], marker="x")
ax = plt.gca()
ax.spines["left"].set_position("zero")
ax.spines["bottom"].set_position("zero")
ax.spines["right"].set_color("none")
ax.spines["top"].set_color("none")
plt.gca().set_aspect("equal", adjustable="box")
plt.show()
```

```python
lattice3 = hz.Lattice(n_max=3, nd=2, basis=basis, no_zero=True)
plt.scatter(lattice3.n[:, 0], lattice3.n[:, 1], marker=".")
ax = plt.gca()
ax.spines["left"].set_position("zero")
ax.spines["bottom"].set_position("zero")
ax.spines["right"].set_color("none")
ax.spines["top"].set_color("none")
plt.gca().set_aspect("equal", adjustable="box")
plt.show()
```
